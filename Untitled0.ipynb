```python
import pandas as pd
import glob

files = sorted(glob.glob("*.xlsx"))
print(files)

nov_file = [f for f in files if "Nov" in f][0]
oct_file = [f for f in files if "Oct" in f][0]

nov = pd.read_excel(nov_file).rename(columns={"Date": "Day"}).melt(id_vars="Day", var_name="Hour", value_name="AQI")
nov["Month"] = "Nov"

octo = pd.read_excel(oct_file).rename(columns={"Date": "Day"}).melt(id_vars="Day", var_name="Hour", value_name="AQI")
octo["Month"] = "Oct"

data = pd.concat([octo, nov], ignore_index=True)
data["AQI"] = pd.to_numeric(data["AQI"], errors="coerce")

print(data.shape)
print("Valid readings:", data["AQI"].notna().sum())
print(data.groupby("Month")["AQI"].agg(["count", "mean", "max"]))
```

```python
import requests
import pandas as pd

url = "https://air-quality-api.open-meteo.com/v1/air-quality?latitude=29.96&longitude=77.55&hourly=pm2_5,pm10&start_date=2025-10-01&end_date=2025-11-30&timezone=Asia/Kolkata"
r = requests.get(url)
js = r.json()
df = pd.DataFrame(js["hourly"])
df["time"] = pd.to_datetime(df["time"])
df["date"] = df["time"].dt.date
df["month"] = df["time"].dt.strftime("%b")
df["hour"] = df["time"].dt.hour
print(df.shape)
print(df["pm2_5"].describe())
```

```python
import matplotlib.pyplot as plt

df["date"] = df["time"].dt.date
df["month"] = df["time"].dt.strftime("%b")
df["hour"] = df["time"].dt.hour

daily = df.groupby("date")["pm2_5"].mean()
print("Worst day:", daily.idxmax(), round(daily.max(), 1))
print("Cleanest day:", daily.idxmin(), round(daily.min(), 1))
print(df.groupby("month")["pm2_5"].mean().round(1))

plt.figure(figsize=(10, 5))
plt.plot(daily.index, daily.values, color="red")
plt.axhline(60, color="black", linestyle="--", label="India 24h standard (60)")
plt.title("Daily Average PM2.5 - Saharanpur (Oct-Nov 2025)")
plt.xlabel("Date")
plt.ylabel("PM2.5 (ug/m3)")
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```
